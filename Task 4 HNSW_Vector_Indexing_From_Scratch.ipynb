# Task 4: In-Memory HNSW Vector Indexing

## What is Vector Retrieval?
Finding the most similar vector (document) to a user query using similarity scores.

```python
import numpy as np

# Sample Document Vectors
documents = {
    "Doc 1 (AI & ML)": np.array([1, 0]),
    "Doc 2 (Deep Learning)": np.array([9, 1]),
    "Doc 3 (Cooking & Recipes)": np.array([0, 1]),
}

user_query = np.array([2, 1]) # User looking for AI topics

def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

# Calculate similarity for all documents
scores = []
for doc_name, vector in documents.items():
    sim = cosine_similarity(user_query, vector)
    scores.append((doc_name, sim))

# Sort documents by highest similarity score
scores.sort(key=lambda x: x[1], reverse=True)

print("Search Results for Query:")
for doc_name, sim in scores:
    print(f"  {doc_name} -> Similarity: {sim:.4f}")
```
