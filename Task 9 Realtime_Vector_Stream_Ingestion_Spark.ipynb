# Task 9: Real-Time Vector Stream Ingestion

## What is Stream Ingestion?
Processing incoming data items one by one, turning text into numbers (vectors), and storing them live.

```python
import numpy as np

vector_database = {}

# Incoming stream of text logs
stream = [
    ("log_1", "User logged in"),
    ("log_2", "Payment completed"),
    ("log_3", "Server restart")
]

# Process stream items
for doc_id, text in stream:
    # Turn text into vector based on character length
    vec = np.array([len(text), len(text.split())], dtype=float)
    vector_database[doc_id] = vec
    print(f"Upserted '{doc_id}' -> Vector: {vec}")

print("\nTotal Stored Vectors:", len(vector_database))
```
