# Task 3: LLM Quantization & Logit Extraction

## 1. Quantization
Converting large 32-bit floating numbers to small 8-bit or 4-bit integers to save memory.

## 2. Entropy (Uncertainty)
Measuring how uncertain the AI is when guessing the next word.

```python
import torch

float_weights = torch.tensor([0.25, -0.50, 0.75, -1.00])
scale_factor = 0.25

# Quantize float -> int8
quantized_int8 = torch.round(float_weights / scale_factor).to(torch.int8)
# Dequantize int8 -> float
dequantized_float = quantized_int8.float() * scale_factor

print("Original Float Weights:", float_weights)
print("Quantized Integers:", quantized_int8)
print("Dequantized Float Weights:", dequantized_float)

logits = torch.tensor([[2.0, 1.0, 0.1]]) # Raw output scores
probs = torch.softmax(logits, dim=-1)     # Convert scores to probabilities

# Entropy formula: -sum(p * log(p))
entropy = -torch.sum(probs * torch.log(probs))

print("\nModel Probabilities:", probs)
print("Model Uncertainty (Entropy):", entropy.item())
```
