```python
import pandas as pd

data = {
    "answer": [
        "I am an MCA student with strong Python skills, extensive project experience in machine learning and data science, and a proven track record of delivering impactful solutions.",
        "I have a deep understanding of Python, with hands-on experience in developing and deploying AI models, contributing to open-source projects, and leading small development teams.",
        "While I am still growing my programming knowledge, I am a dedicated learner with a strong foundational understanding and a keen interest in expanding my skills rapidly.",
        "I am highly motivated, possess exceptional problem-solving abilities, and am a quick learner who thrives in challenging environments, committed to continuous self-improvement.",
        "I have limited professional experience, but I have actively engaged in several personal projects and online courses to build a strong practical skillset in data analysis and programming.",
        "My core strength lies in translating complex business problems into actionable data-driven solutions, leveraging advanced Python libraries and statistical modeling.",
        "I am a proactive team player who consistently seeks opportunities to contribute beyond my immediate responsibilities, ensuring project success and fostering a collaborative environment.",
        "I am passionate about leveraging technology to solve real-world problems and am constantly exploring new tools and methodologies to enhance my capabilities."
    ],
    "score": [9.8, 9.5, 3.0, 9.0, 4.5, 9.7, 9.2, 9.6]
}

df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import pickle

# Dataset
answers = [
    "I am an MCA student with strong Python skills and project experience",
    "I know Python and have worked on AI projects",
    "I don't know much about programming",
    "I am hardworking and a quick learner",
    "I have no experience"
]

scores = [9, 8, 2, 7, 1]

# Convert text to vectors
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(answers)

# Train model
model = LinearRegression()
model.fit(X, scores)

# Save model
pickle.dump(model, open("interview_model.pkl", "wb"))
pickle.dump(vectorizer, open("vectorizer.pkl", "wb"))

print("Model trained successfully!")
```

```python
import pickle

# Load model
model = pickle.load(open("interview_model.pkl", "rb"))
vectorizer = pickle.load(open("vectorizer.pkl", "rb"))

questions = [
    "Tell me about yourself. What motivated you to apply for this role?",
    "What are your greatest strengths, and how do they align with this position?",
    "Why should we hire you? What unique value can you bring to our team?",
    "Can you describe a challenging project you've worked on and how you overcame obstacles?",
    "Where do you see yourself in five years, and how does this role fit into your career aspirations?"
]

print("===== AI Interview Coach ====")

for question in questions:
    print("\nQuestion:", question)

    answer = input("Your Answer: ")

    answer_vector = vectorizer.transform([answer])

    score = model.predict(answer_vector)[0]

    score = max(0, min(10, round(score, 1)))

    print("Predicted Score:", score, "/10")

    if score >= 8.5:
        print("Excellent Answer! Very impressive.")
    elif score >= 6.5:
        print("Good Answer. Consider adding more specific examples or quantitative results.")
    elif score >= 4.0:
        print("Needs Improvement. Focus on clearly articulating your skills, projects, and achievements relevant to the role.")
    else:
        print("Significant Improvement Needed. Review common interview techniques and practice structuring your responses.")
```
