# Task 1: Vectorized Scaled Dot-Product Attention

## What is Attention?
Attention helps models focus on relevant words in a sentence.
- **Query (Q)**: What we are looking for.
- **Key (K)**: What each word offers.
- **Value (V)**: The actual information stored for each word.

Formula: `Output = Softmax( (Q @ K^T) / sqrt(d_k) ) @ V`

```python
import numpy as np

# Create simple sample inputs for 2 words
Q = np.array([[1, 4], [0, 1]])  # Query matrix
K = np.array([[1, 4], [0, 1]])  # Key matrix
V = np.array([[10, 20], [30, 40]]) # Value matrix

d_k = Q.shape[-1] # dimension size (2)

# Multiply Query and Key (Transpose of K) and scale
scores = np.dot(Q, K.T) / np.sqrt(d_k)

# Apply Causal Mask (hide future words with negative infinity)
mask = np.tril(np.ones_like(scores))
scores_masked = np.where(mask == 0, -1e9, scores)

# Turn scores into probabilities using Softmax
exp_scores = np.exp(scores_masked - np.max(scores_masked, axis=-1, keepdims=True))
weights = exp_scores / np.sum(exp_scores, axis=-1, keepdims=True)

# Multiply attention weights with Value matrix
output = np.dot(weights, V)

print("1. Attention Scores:", scores)
print("2. Softmax Attention Weights:\n", np.round(weights, 2))
print("3. Final Attention Output:\n", np.round(output, 2))
```
