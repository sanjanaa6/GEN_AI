# Task 10: DDPM Diffusion Model

## Forward Process
Adding noise to an image step by step.

## Reverse Process
Removing noise step by step to recover the clean image.

```python
import torch

# Clean image values (1.0)
clean_image = torch.tensor([1.0, 1.0, 1.0])

# Forward process: Add 50% noise
noise = torch.randn(3)
noisy_image = 0.5 * clean_image + 0.5 * noise

# Reverse process: Subtract estimated noise to restore clean image
estimated_noise = noise
restored_image = (noisy_image - 0.5 * estimated_noise) / 0.5

print("1. Clean Image:", clean_image)
print("2. Noisy Image (Forward):", noisy_image)
print("3. Restored Image (Reverse):", restored_image)
```
