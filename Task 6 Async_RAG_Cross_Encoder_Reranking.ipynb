# Task 6: Async RAG Pipeline & Reranking
## What is Async RAG?
1. **Retrieval**: Quickly find candidate documents matching words.
2. **Reranking**: Select the best document.
3. **Async**: Process multiple user queries at the exact same time.

```python
import asyncio

documents = [
    "Python is a great programming language.",
    "Transformers use self-attention to process text.",
    "RAG combines retrieval with text generation."
]

async def search_and_rerank(query):
    # Filter candidates
    candidates = [doc for doc in documents if any(w in doc.lower() for w in query.lower().split())]
    if not candidates:
        candidates = documents

    # Pick best matching candidate
    best_doc = max(candidates, key=len)

    # Simulate async delay
    await asyncio.sleep(0.1)
    return f"Query: '{query}' -> Best Result: '{best_doc}'"

async def main():
    # Run multiple queries asynchronously
    results = await asyncio.gather(
        search_and_rerank("python programming"),
        search_and_rerank("attention mechanism")
    )
    for r in results:
        print(r)

await main() if '__file__' not in globals() else asyncio.run(main())
```
