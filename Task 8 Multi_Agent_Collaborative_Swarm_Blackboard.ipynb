# Task 8: Multi-Agent Swarm with Shared Blackboard

## What is a Blackboard Architecture?
A shared memory dictionary where multiple AI agents write their outputs safely using thread locks.

```python
import threading

blackboard = {} # Shared memory store
lock = threading.Lock()

def coder_agent():
    with lock:
        blackboard["code"] = "def add(a, b): return a + b"
        print("Coder Agent: Code created.")

def reviewer_agent():
    with lock:
        blackboard["review"] = "Code review PASSED."
        print("Reviewer Agent: Code reviewed.")

# Start parallel threads
t1 = threading.Thread(target=coder_agent)
t2 = threading.Thread(target=reviewer_agent)

t1.start()
t2.start()
t1.join()
t2.join()

print("\nFinal Shared Blackboard Contents:\n", blackboard)
```
