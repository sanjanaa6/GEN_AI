# Task 2: Custom BPE Tokenizer & Causal LM

## What is BPE Tokenization?
It counts character pairs and merges the most frequent pair together into subwords.

## What is a Causal LM?
A language model that predicts the next token while hiding future tokens with a mask.

```python
from collections import Counter
import torch
import torch.nn as nn


text = "hug hugging hugger"
words = [list(w) for w in text.split()]

# Count pairs
pairs = Counter()
for w in words:
    for i in range(len(w) - 1):
        pairs[(w[i], w[i+1])] += 1

most_common_pair = pairs.most_common(1)[0][0]
print("Most Frequent Character Pair:", most_common_pair)

# Simple model with 10 tokens vocabulary and embedding size 4
input_tokens = torch.tensor([[1, 2, 3]])  # Sequence of 3 words

embedding = nn.Embedding(num_embeddings=10, embedding_dim=4)
linear = nn.Linear(4, 10)

# Causal Mask (Lower Triangular Matrix to hide future tokens)
mask = torch.tril(torch.ones(3, 3))
print("Causal Mask Matrix:\n", mask)

embedded_words = embedding(input_tokens)
masked_features = embedded_words * mask.unsqueeze(0).unsqueeze(-1)
logits = linear(masked_features)

print("Output Logits Shape:", logits.shape)
```
