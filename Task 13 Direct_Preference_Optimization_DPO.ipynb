# Task 13: Direct Preference Optimization (DPO)

## What is DPO?
Fine-tuning an AI by showing it pairs of answers: a **Chosen (good)** answer and a **Rejected (bad)** answer.

```python
import torch
import torch.nn.functional as F

# Model scores for chosen (good) vs rejected (bad) responses
score_chosen = torch.tensor([2.5])   # High score preferred
score_rejected = torch.tensor([0.5]) # Low score preferred

# DPO Loss: -log(sigmoid(score_chosen - score_rejected))
margin = score_chosen - score_rejected
loss = -F.logsigmoid(margin)

print("1. Score Difference (Margin):", margin.item())
print("2. DPO Preference Loss:", loss.item())
```
