# Task 15: Enterprise LLM Gateway with Rate Limiting

## What is an LLM Gateway?
1. **Rate Limiter**: Rejects requests if user sends too many (e.g., max 3 requests).
2. **Fallback Governance**: If the primary AI model fails (server error), automatically switch to a backup model.

```python
request_count = 0
MAX_LIMIT = 3

def llm_gateway(prompt):
    global request_count
    request_count += 1

    # Rate Limiting Check
    if request_count > MAX_LIMIT:
        return "429 Error: Rate Limit Exceeded (Too Many Requests!)"

    # Model Execution with Fallback
    try:
        if request_count == 2:
            raise Exception("Primary Model 503 Server Error")
        return f"Primary LLM Model Answer for '{prompt}'"
    except Exception as e:
        print(f"  [Gateway Alert]: {e}. Switching to backup model...")
        return f"Backup Secondary LLM Model Answer for '{prompt}'"

# Test 4 requests
for i in range(1, 5):
    response = llm_gateway(f"Question {i}")
    print(f"Request {i}: {response}\n")
```
