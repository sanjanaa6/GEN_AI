# Task 14: Custom SwiGLU Activation Function

## What is SwiGLU?
SwiGLU is an activation function calculated as:
`SwiGLU(x, g) = (x * sigmoid(x)) * g`

```python
import torch

def swiglu(x, g):
    # Compute Swish part: x * sigmoid(x)
    swish_x = x * torch.sigmoid(x)
    # Multiply by gate g
    return swish_x * g

# Sample inputs
x = torch.tensor([1.0, -2.0, 3.0])
g = torch.tensor([0.5,  0.5, 0.5])

output = swiglu(x, g)

print("Input x:", x)
print("Gate g:", g)
print("SwiGLU Output:", output)
```
