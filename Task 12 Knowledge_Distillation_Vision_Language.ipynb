# Task 12: Knowledge Distillation

## What is Knowledge Distillation?
Teaching a small "Student" model to mimic the outputs of a large "Teacher" model.

```python
import torch
import torch.nn as nn

# Big Teacher Model (Fixed / Frozen output)
teacher_output = torch.tensor([[8, 2, 0]])

# Small Student Model
student_layer = nn.Linear(3, 3)
input_data = torch.tensor([[1, 5, 0.1]])

student_output = student_layer(input_data)

# Distillation Loss (Difference between Student and Teacher)
loss_fn = nn.MSELoss()
loss = loss_fn(student_output, teacher_output)

print("Teacher Output Target:", teacher_output)
print("Student Output Prediction:", student_output.detach())
print("Distillation Loss:", loss.item())
```
