# Task 7: LoRA Fine-Tuning

## What is LoRA?
Instead of retraining a huge model, we freeze the original weights and train two small low-rank matrices `A` and `B`.

```python
import torch
import torch.nn as nn

# Big Original Base Layer (FROZEN - no training)
base_layer = nn.Linear(10, 10)
for p in base_layer.parameters():
    p.requires_grad = False # Freeze

# LoRA Adapter Matrices: A (10x2) and B (2x10) - TRAINABLE
rank = 2
lora_A = nn.Parameter(torch.randn(10, rank))
lora_B = nn.Parameter(torch.zeros(rank, 10))

# Compute final output
x = torch.randn(1, 10)
base_out = base_layer(x)
lora_out = x @ lora_A @ lora_B

final_output = base_out + lora_out

print("Is Original Layer Frozen?", not base_layer.weight.requires_grad)
print("Is LoRA Matrix A Trainable?", lora_A.requires_grad)
print("Final Output Shape:", final_output.shape)
```
